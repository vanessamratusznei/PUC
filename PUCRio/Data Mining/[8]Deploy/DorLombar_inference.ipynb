```
# 1. Carregaremos o modelo treinado
import joblib
filename = 'spine_model.pkl'
model = joblib.load(open(filename, 'rb'))
```

```
# 2. Carregaremos novos dados a serem inferidos
import pandas as pd
new_data = pd.read_csv('Dataset_spine_unknown.csv')
new_data.head()
```

```
# 2.1, se eu tivesse um pré-processamento de normalização
# new_data = scaler.transform(new_data)
```

```
# 3. Faremos as inferências
inferences = model.predict(new_data)
print(inferences)
```

```
# exportar os resultados
new_data['inferences'] = inferences
new_data.head(len(new_data))
```
