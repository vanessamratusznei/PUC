### Encoding da Variável Occupation

###Encoding da Variável BMI Category e Sleep Disorder

```python
def BMI_Category_encoding(x):
    encoding=BMI_category_list.index(x)
    #if x=='Overweight'
    #  encoding=10

    return encoding

Sleep_Disorder_encoding_list=list(df['Sleep Disorder'].unique())
def Sleep_Disorder_encoding(x):
    encoding=Sleep_Disorder_encoding_list.index(x)
    return encoding
```

```python
df['Sleep_Disorder_encoding'] = df.apply(lambda x: Sleep_Disorder_encoding(x['Sleep Disorder']), axis=1)
```

### Encoding da Variável Gender

*Vamos gerar uma coluna nova para cada variável!
*(One hot encoding)

```python
Gender_list=list(df['Gender'].unique())
print(Gender_list)

for gen in Gender_list:
  df[gen] = np.where(df['Gender']==gen, 1, 0)
df.head()
```

##Normalizacao

*Columns : Sleep Duration ; Physical Activity Level; Heart Rate ;Daily Steps ;

```python


print(df['Sleep Duration'].max())
print(df['Sleep Duration'].min())


print(df['Physical Activity Level'].max())
print(df['Physical Activity Level'].min())


print(df['Heart Rate'].max())
print(df['Heart Rate'].min())


print(df['Daily Steps'].max())
print(df['Daily Steps'].min())
```

```python
def Sleep_Duration_normalizacao(x):
    x=(x-5.8)/(8.5-5.8)
    return x
def Physical_Activity_Level_normalizacao(x):
    x=(x-30)/( 90-30)
    return x
def Heart_Rate_normalizacao(x):
    x=(x-65)/(86-65)
    return x
def Daily_Steps_normalizacao(x):
    x=(x-3000)/(10000-3000)
    return x
```

```python
df['Sleep Duration_normal'] = df.apply(lambda x: Sleep_Duration_normalizacao(x['Sleep Duration']), axis=1)
df['Physical Activity Level_normal'] = df.apply(lambda x: Physical_Activity_Level_normalizacao(x['Physical Activity Level']), axis=1)
df['Heart Rate_normal'] = df.apply(lambda x: Heart_Rate_normalizacao(x['Heart Rate']), axis=1)
df['Daily Steps_normal'] = df.apply(lambda x: Daily_Steps_normalizacao(x['Daily Steps']), axis=1)
```

```python
df.head()
```

## Validar variavel Blood Pressure
* A variável está invalida pro modelo. Vamos realizar um encoding valido


```python
def fix_variavel(x):
    b1=x.split('/')[0]
    b2=x.split('/')[1]
    bp_encoding=int(b2)/int(b1)
    return bp_encoding

df['Blood Pressure_normal']= df.apply(lambda x: fix_variavel(x['Blood Pressure']), axis=1)
```

## Separar em treino e teste

```python

from sklearn.model_selection import train_test_split
# Split the data into training and testing sets

train_features, test_features, train_labels, test_labels = train_test_split(features_np, labels_np, test_size = 0.25, random_state = 42)
```

# Previsão
